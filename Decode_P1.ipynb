```python
import random
from datetime import datetime

print("""
========== Chat Bot ==========
Type 'bye' or 'exit' to end chat
==============================
""")

responses = {
    "hello": ["Hi there!", "Hello!", "Hey!"],
    "hi": ["Hi Dear!", "Hello!", "Hey there!"],
    "how are you": [
        "I'm good!",
        "Doing great!",
        "Nice to talk with you!"
    ],
    "name": [
        "I'm a Python chatbot.",
        "My name is PyBot."
    ],
    "help": [
        "Try saying hello.",
        "You can ask my name or the time."
    ]
}


def get_time():
    now = datetime.now()
    return now.strftime("%H:%M:%S")


def chatbot_response(user_input):

    if user_input in ["bye", "exit"]:
        return "Goodbye, See you later!"

    elif "time" in user_input:
        return f"Current time is: {get_time()}"

    elif user_input in responses:
        return random.choice(responses[user_input])

    else:
        return "Sorry, I don't understand."


while True:

    user_input = input("You: ").strip().lower()

    response = chatbot_response(user_input)

    print("Bot:", response)

    if user_input in ["bye", "exit"]:
        break
```

```python
import random
from datetime import datetime

print("""
========== Smart Chat Bot ==========
Type 'bye' or 'exit' to end chat
===================================
""")

responses = {

    ("hello", "hi", "hey", "good morning", "good evening"):
    [
        "Hello!",
        "Hi there!",
        "Hey!"
    ],

    ("how are you", "how are you doing", "are you fine"):
    [
        "I'm doing great!",
        "I'm good, thanks!",
        "Awesome!"
    ],

    ("what is your name", "your name", "name"):
    [
        "My name is PyBot.",
        "I'm a Python chatbot."
    ],

    ("help", "can you help me", "i need help"):
    [
        "Of course!",
        "Tell me what you need."
    ],

    ("thanks", "thank you", "thx"):
    [
        "You're welcome!",
        "Anytime!"
    ]
}


def get_time():
    return datetime.now().strftime("%H:%M:%S")


def chatbot_response(user_input):

    if user_input in ["bye", "exit"]:
        return "Goodbye!"

    if "time" in user_input:
        return f"The time is {get_time()}"

    for keywords, replies in responses.items():

        if user_input in keywords:
            return random.choice(replies)

    return "Sorry, I don't understand."


while True:

    user_input = input("You: ").strip().lower()

    response = chatbot_response(user_input)

    print("Bot:", response)

    if user_input in ["bye", "exit"]:
        break
```
